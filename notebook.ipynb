```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
```

```python
import os

filename = 'healthcare_dataset.csv'
root_dir = '/workspaces/Synthetic-Healthcare-Data-Analysis-/'

for root, dirs, files in os.walk(root_dir):
    if filename in files:
        path = os.path.join(root, filename)
        print(path)
    print("file name not found.")
```

```python
df = pd.read_csv(path) 
df.head(10)
```

# Data Transformation

- Lower case name
- Sort by name
- Lower case column name & Replace space with _
- Remove any duplications by subset of columns: name, date of admission, doctor, discharge date
- Replace name to index number as a unique id

### NOTE

The data doesn't contain person identity, therefore cannot provide unique id at this point with the given informations. 

```python
df['Name'] = df['Name'].str.lower()
df_sorted = df.sort_values(by = ['Name'])
df_sorted
```

```python
df_sorted.columns = df_sorted.columns.str.lower().str.replace(" ", "_")
```

```python
df_sorted[df_sorted.duplicated()]
```

```python
df_clean = df_sorted.drop_duplicates(subset = ['name', 'medical_condition', 'date_of_admission', 'doctor', 'discharge_date'])
df_clean
```

```python
df_clean = df_clean.reset_index().drop(labels = 'index', axis = 1)
df_clean.head(10)
```

```python
df_clean[df_clean.duplicated()]
```
